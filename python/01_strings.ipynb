## 2. A string is text in quotes
Python calls a piece of text a **string**. You write one in quotes. A **variable** is a name you give a value so you can use it again: `=` means *"from now on, `name` means this"*.

````python
name = "gArCiA"
print(name)
print(len(name))      # len() counts the characters
````

Spaces and tabs are characters too, and `print` doesn't show them. Some last names in the file are nothing but spaces and a tab. Put brackets around one with `+` (joining strings end to end), and you can see where it starts and ends:

````python
blank = "    \t    "
print("[" + blank + "]")
print(len(blank))
````

## 3. Things a string can do
Each string comes with **methods**: actions you call with a dot and a pair of brackets. The brackets mean *"do it now"*.

````python
print(name.upper())
print(name.lower())
print(name.title())
print("[" + blank.strip() + "]")     # strip() removes spaces and tabs from both ends
````

One thing to know: a method **never changes** the string. It hands back a new one. `name` is still exactly what it was:

````python
print(name)
````

You can also chain methods. Python does them left to right: `blank.strip().upper()` strips first, then capitalizes what's left.

## 4. Asking a question with `==`
`=` gives a value a name. `==` (two equals signs) **asks** whether two values are the same, and the answer is `True` or `False`. Exact matches like Q3's, done as written, use `==`:

````python
print("Garcia" == "GARCIA")
print("Garcia".upper() == "GARCIA")
````

Capitalize both sides first, and the capitals stop mattering.

## 5. Putting values into text: f-strings
Put an `f` before the quotes, and anything in `{curly brackets}` is worked out and dropped into the text:

````python
first, last = "maria", "GARCIA"
print(f"{last.title()}, {first.title()}")
print(f"{name} has {len(name)} letters")
````

## 6. Your first function
Part 0 called a function a **recipe**. Now you'll write one. Here's a tiny one:

````python
def shout(text):
    """The text in capitals, with an exclamation mark."""
    return text.upper() + "!"

shout("kent")
````

- `def shout(text):` starts a recipe called `shout`. Whatever you give it is called `text` inside.
- The indented lines are the recipe. Python knows where it ends because the indenting stops.
- The line in triple quotes says what it does (a *docstring*). Every function in course 1 has one.
- `return` hands the result back.
